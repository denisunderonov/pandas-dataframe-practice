## Пункт второй

Создай сэмпл из 200 новых наблюдений, используя random_state = 21.
В сэмпле не должно появиться новых комбинаций car number, make и model, чтобы набор данных оставался согласованным.
На столбцы refund и fines ограничений нет: можно случайно выбирать их значения и назначать любому номеру автомобиля.

```python
sample = ds.sample(n=200,random_state=21)
sample
```

## Пункт третий

Конкатенируй сэмпл с исходным DataFrame и получи новый DataFrame concat_rows.

```python
concat_rows = pd.concat([ds, sample])
concat_rows
```

## Пункт четвертый 

Создай Series с именем Year со случайными целыми числами от 1980 до 2019.
Перед генерацией лет вызови np.random.seed(21).
Конкатенируй Series с DataFrame и назови получившийся DataFrame fines.


```python
np.random.seed(21)

years = pd.Series(np.random.randint(1980,2020, size=len(concat_rows)), name="Year", index=concat_rows.index)
fines = pd.concat([concat_rows, years], axis=1)
fines
```

## Пункт пятый

Создай новый DataFrame с номерами автомобилей и их владельцами.
Возьми самые популярные фамилии в США (файл surname.json в папке datasets).
Создай новую Series с фамилиями: без спецсимволов (запятых, скобок и т. п.); количество должно совпадать с числом уникальных номеров автомобилей в сэмпле (используй random_state = 21).
Создай DataFrame owners с колонками CarNumber и SURNAME.

```python
surnames = pd.read_json("data/surname.json")
car_numbers = fines["CarNumber"].drop_duplicates()
selected_surnames = surnames["SURNAME"].sample(n=car_numbers.nunique(), random_state=21).reset_index(drop=True)
owners = pd.DataFrame({
    "CarNumber": car_numbers.reset_index(drop=True),
    "SURNAME":selected_surnames
})
owners
```

## Пункт шестой

Добавь ещё пять наблюдений в fines (придумай свои CarNumber и т. п.).

```python
new_rows = pd.DataFrame({
    "CarNumber": [
        "A001AA77RUS",
        "B002BB77RUS",
        "C003CC77RUS",
        "E004EE77RUS",
        "K005KK77RUS"
    ],
    "Refund": [1, 0, 2, 1, 0],
    "Fines": [1500.0, 2500.0, 500.0, 3000.0, 750.0],
    "Make": ["Toyota", "BMW", "Ford", "Kia", "Hyundai"],
    "Model": ["Camry", "X5", "Focus", "Rio", "Solaris"],
    "Year": [2015, 2018, 2012, 2019, 2017]
})

fines = pd.concat([fines, new_rows], ignore_index=True)
fines.tail()
```

## Пункт седьмой

Удали последние 20 наблюдений из owners и добавь три новых, отличных от добавленных в fines.

```python
owners = owners.iloc[:-5]
new_owners = pd.DataFrame({
    "CarNumber": [
        "M111MM99RUS",
        "P222PP50RUS",
        "T333TT78RUS"
    ],
    "SURNAME": [
        "MILLER",
        "WILSON",
        "ANDERSON"
    ]
})

owners = pd.concat([owners, new_owners], ignore_index=True)
owners.tail()
```

## Пункт восьмой, соединения таблиц

```python
#1. Только номера, которые есть в обоих DataFrame
both = fines.merge(owners, on="CarNumber", how="inner")

#2. Все номера из обоих DataFrame
all_numbers = fines.merge(owners, on="CarNumber", how="outer")

#3. Все номера из fines
from_fines = fines.merge(owners, on="CarNumber", how="left")

#4. Все номера из owners
from_owners = fines.merge(owners, on="CarNumber", how="right")
```
